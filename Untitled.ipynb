## Reverse Array

### Approach 1 

#### Array Reverse using extra array. TC- O(n), SC- O(n)

```python
def reverse(arr):
    reverse_array = arr[::-1]
    return reverse_array

reverse([1,2,3,4])
```

### Approach 2

#### Array reverse using 2 pointer. TC - O(n), SC - O(1)

```python
def reverse_point(arr, start, end):

    while start < end:
        arr[start], arr[end] = arr[end], arr[start]
        start += 1
        end -= 1
A = [1,2,3,4,5,6]
reverse_point(A,0,5)
print(A)
```
